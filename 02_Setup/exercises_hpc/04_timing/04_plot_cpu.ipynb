```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('04_out_pool_fire_cpu.csv')
print(data.head)
```

```python
# do not consider the rank and the total time
reduced_data = data.drop(columns=['Rank', 'Total T_USED (s)'])

# plot the data with the column labels at the x-axis
reduced_data.transpose().plot(kind='bar', figsize=(8,4))

plt.legend(title='MPI rank')
plt.ylabel('time / s')
plt.tight_layout()
plt.savefig('benchmark_cputime.png', dpi=300)
```
